## Graficación de una onda viajera con funciones y librerías científicas

Se define una función en Python que modela una onda sinusoidal viajera del curso de Física. Se hace uso de las 
librerías **NumPy** para el manejo de arreglos matemáticos y **Matplotlib** para la visualización de la onda en distintos 
instantes de tiempo.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def onda_viajera(x, t): #y(x,t)=0,350sen(10*pi*t-3*pi*x+pi/4)
    A = 0.350
    k = 3 * np.pi
    omega = 10 * np.pi
    phi = np.pi / 4
    return A * np.sin(omega * t - k * x + phi)
```

```python
x = np.linspace(0, 2, 1000)

plt.plot(x, onda_viajera(x, 0), label="t = 0 s")
plt.plot(x, onda_viajera(x, 0.05), label="t = 0.05 s")
plt.title("Onda viajera y(x,t) = 0.350 sen(10πt − 3πx + π/4)")
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.legend()
plt.grid()
plt.show()
```
