## Uso de elif en los condicionales

La estructura *elif* resulta de una contracción entre *else + if*. Sirve para realizar evaluaciones más complejas de las condiciones.

 **Ejemplo**

> Definir **x** e **y** como dos variables numéricas enteras. Si **x** e **y** son iguales, entonces se obtendrá el cuadrado de cualquiera de ellas y se imprimirá el resultado en pantalla. En cambio, si **x** es mayor que **y**, se obtendrá la diferencia **x-y**. De lo contrario, se obtendrá la suma.


````python
# Definimos x
x= 23

#Definimos y
y=-14

if x == y:
    print("El cuadrado de x es ", x**2)
elif  x > y:
    print("La diferencia entre x e y es ", x-y)
else:
    print("La suma entre x e y es ", x+y)
````


## La Serie de Fibonacci

La Serie de Fibonacci es una secuencia infinita de números naturales. La misma inicia con 1 y 1. Cualquier otro término se obtiene sumando los dos números inmediatamente anteriores. Para los primeros términos la serie toma los valores:

**1, 1, 1+1=2, 1+2=3, 2+3=5...**

En términos matemáticos,cada elemento $r_{i}$   se obtiene como:
       $r_{i}=r_{i-2}+r_{i-1}$
En dónde (i-2) representa el valor que tenía r hace dos posiciones y (i-1) el valor una posición anterior.
               
               
         
##  Ejercicio

````python
#------------------------------#
#  Primero vamos a generar una serie de Fibonacci
# en la que (arbitrariamente) asignamos como un valor máximo n
# de 10
#------------------------------#
#- Máximo valor que puede tomar r (i-2)
n = 10

#- Definimos una variable a que representará a r(i-2)
# y le asginamos el valor 0
a = 0

#- Definimos una variable b que representará a r(i-1)
# y le asginamos el valor 1
b = 1

# Creamos una lista vacía que almacenará los resultados de la serie
l0 = [] 

# Empleamos un bucle condicional indeterminado, 
# puesto que no sabemos en que punto se alcanzará 
# a se aproximará al valor máximo n
while a < n:
    
    #Imprimir valores iniciales
    print(a, b)
    
    # Actualizar a guardando el valor de b
    a = b 
    
    # Actualizar el valor de b, como la suma de a  y b
    b = a+b 
    
    # Imprimir los nuevos valores  
    print(a,b)
    
    # Guardar los valores en la lista
    l0.append(a+b)
    print ('La serie contiene:', l0)
````

````python
# Adoptamos el código anterior en función
# para que funcione para cualquier valor máximo de n

def fibonnaci(n):
    """ 
      Calcula la serie de Fibonnaci, en la cual cada elemento r se obtiene de los elementos 
      inmediatamente anteriores r(i-2), r(i-1). El cálculo se detiene cuando r(i-2) > n
     #----#
     # Entrada
         n : numérico. Límite superior del término r(i-2).
     # Salida:
        l0: lista. Serie de Fibonacci
     """
    a, b = 0, 1
    l0 = [] 

    while a < n:
        a, b = b, a+b
        l0.append(a+b)
    
    return l0

#-------------------#
fibonnaci?
````
