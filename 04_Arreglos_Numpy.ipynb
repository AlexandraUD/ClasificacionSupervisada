# Arreglos (array) de NumPy

Pero qué es un *array*? 

Un *array* (o arreglo) es una estructura para almacenar datos. Tenemos que pensar en estos arreglos como matrices o grillas, en dónde cada celda almacena un dato. Estos arreglos pueden ser *n*-dimensionales, por ejemplo: 1D, 2D, 3D..



Cada dimensión es un eje.

**Características de los arreglos**:

 Son rectangulares.

 Todos los elementos almacenados dentro del arreglo son del mismo tipo: todos enteros, todos decimales, etc.

 Una vez creados, no es posible modificar su forma.

## Crear Arreglos


```python
#-------------#
# Desde listas
#--------------#
#Crear lista
l0 = [ 3, 1.4, 5]

# convertir a arreglo
a0 = np.asarray(l0)
print(type(a0), a0)
```

```python
#-------------#
# Desde tuplas
#-------------#
t0 = (8,9)

# convertir a arreglo
a1 = np.asarray(t0)
print(type(a1), a1)
```

```python
#------------#
# Con funciones generadoras
#-------------#
# np.linspace genera un arreglo de 80 números en total, 
# interpolados linealmente entre 1 y 23
a2 = np.linspace(1, 23, num=80)
print('Arreglo generado con np.linspace \n', a2)
print('*'*50)

#-------------#
# np.arange genera un arreglo con números entre dos extremos
# a un cierto paso
a3 = np.arange(1, 23, 1)
print('Arreglo generado con np.arange \n', a3)
print('*'*50)

#--------------#
# np.zeros genera un array vacío de cierta dimensión
a4 = np.zeros((3,2), dtype=float)
print('Arreglo generado con np.zeros \n', a4)
print('*'*50)

#--------------#
# np.ones genera un array lleno de 1 unos
a5 = np.ones( (3, 2), dtype= float)
print('Arreglo generado con np.ones \n',a5)
print('*'*50)
```

# Concatenación de Arreglos

En las siguientes líneas aprenderemos a concatenar y apilar arreglos.

Para concatenatar y apilar los arreglos deben tener la misma dimensión así como elementos en cada dimensión.

```python
# Crear dos arreglos con las mismas dimensiones
A1 = np.arange(5).reshape(1, 5) 
A2 = np.zeros(shape=(1, 5)) 

#
print(A1,A2)
```

```python
# Concatenar por el axis 0 - Apilar filas
A3 = np.concatenate((A1, A2), axis= 0)
print(A3)
```

```python
# np.vstack equivale a  np.concatenate((A1, A2), axis= 0)
np.vstack((A1, A2))
```

```python
# Concatenar por el axis 1 - Cada elemento es una columna
A4 = np.concatenate((A1, A2), axis= 1)
print(A4)
```

```python
#np.hstack((A1, A2)) equivale np.concatenate((A1, A2), axis= 1)
np.hstack((A1, A2))
```

##  Desafío
Escribe un algoritmo que imprima 1 a 1, cada elemento de un array. Además en cada paso imprimirá también el valor de los índices. Implemente para ello dos bucles for.

```python
# Define un arreglo de 6 elementos, 
# con forma de 3 filas y dos columnas
A = np.arange(6).reshape(3,2)
print( A)
print('*'*50)

#-----------#
# Iterar sobre cada elemento del arreglo
#-------------#
# i es el índice de las columnas
# j es el índice de las filas
#--------------#

# Primer bucle para
# Iterar sobre las filas
for j in range(0, A.shape[0]):
    
    # Segundo bucle para
    #Iterar sobre las columnas
    for i in range(0, A.shape[1]):
        
        print(A[j,i])
        
        print('Fila: ', j, 'Columna: ', i)
        
        print('+'*50)
    
```

# Operaciones con arrays

Hay muchas operaciones que podemos realizar con arreglos de datos, puesto que todo el paquete python fue diseñado y optimizado para tal fin. No obstante, aquí nos concentraremos en algunas operaciones que se dividen en tres categorías:

* **Operaciones de elemento a elemento (element-wise)**

Cuando intervienen dos arreglos de las mismas dimensiones. Observa las sumas de los dos arreglos.



* **Operaciones de Broadcasting**

Cuando intervenien arreglos de distintas dimensiones. 



* **Operaciones de reducción**
Cuando se reduce la dimensionalidad del array, y se extrae, por ejemplo, un único número. En el caso de la figura, la reducción se ha realizando sumando todos los elementos disponibles en el array.




```python
# Definir dos arreglos 
A=np.arange(6).reshape(3,2)
B=np.random.randint(1,100,size=(3,2))

#Suma
print(A)
print('+')
print(B)
print('-'*10)
print( A + B)
```

Para estas operaciones aritméticas elemento a elemento emplearemos los mismos operadores: 
* Suma **+** 
* Resta **-**
* División **/**

En general, consiste en transformar y alinear los ejes de los distintos arreglos hasta que coincidan en sus dimensiones, y el cálculo pueda ser realizado. Algunas situaciones en las que ocurre:
 
  


 (Figura [Python Handbook](https://jakevdp.github.io/PythonDataScienceHandbook/06.00-figure-code.html#Broadcasting) )
