* Escribe un algoritmo que calcule el seno, coseno y tangente de una lista de ángulos.

> **Lista de ángulos (en grados)**: [ 1, 23.4, 38, 51.2, 81.1 ]

* Las tabla será exportada como *.csv* y también debe ser convertida a un array y guardada en formato *.npy*

````python
import os
import math
import numpy as np
from numpy import *
import matplotlib.pyplot as plt
import pandas as pd
````

 Crea un contenedor que almacena los ángulos y conviertela de grados a radianes

 Puedes calcular las funciones seno o coseno aplicando el resultado a una tabla o aplicando el resultado a un arreglo.

````python
ag = [ 1, 23.4, 38, 51.2, 81.1 ]
ar = np.radians(ag)
s = np.sin(ar)
c = np.cos(ar)
t = np.tan(ar)

print (type(ar))
print ('-'*10)

atributo = ["ang_grad", "ang_rad", "sin", "cos", "tan"]
indice = ["a_g", "a_r", "sin", "cos", "tan"]

df0 = pd.DataFrame({
    "ang_grad": ag,
    "ang_rad":  ar,
    "sin":      s,
    "cos":      c,
    "tan":      t
}, index = indice)

print(df0)
````

 Cuando tengas tu tabla lista, con todas las columnas indicadas, exporta el resultado a *.csv*. Para exportar a *npy*, deberás convertir primero la tabla a un arreglo de numpy y luego guardae el arreglo con extensión *.npy*

````python
# exportar a csv
df0.to_csv('angulos.csv')

print('El elemento 0,0 del dataframe es: ', df0.iloc[0, 0])

# dataframe to array
a0 = df0.iloc[:].values
print ('-'*10)

print('El dataframe a arreglo es: ', a0)
print ('-'*10)

print ('Su tamaño es: ', np.shape (a0))
print ('-'*10)

print ('Su tipo es: ',type (a0))

# guardar arreglo como .npy
np.save('Arreglo_Angulos.npy', a0)
print ('-'*10)

# cargar .npy
data= np.load('Arreglo_Angulos.npy')
print(data)
````
